```python
import numpy as np
```

```python
numbers = np.array([10,20,30,40,50])
results  =  numbers * 4
print(results)
print(results[(results > 10) & (results < 200)])
```

```python
results > 40 # checking the results which gives the boolean value
```

```python
price = np.array([100,200,300,400])
quantity = np.array([2,4,6,8])
total_amount = price * quantity
print(total_amount)
```

```python
# Broadcasting

a = np.array([1,2,3,4,5])
b = 3

print(a * b)
print(a + b)
print(a - b)
```
