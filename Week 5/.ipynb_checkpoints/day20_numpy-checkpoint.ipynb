```python
# using numpy
numbers = np.array([1,2,3,4,5])
results = numbers * 10
print(results)
print(type(results))
```

```python
# Using python
numbers = [1,2,3,4,5]
results = []
for i in numbers:
    results.append(i * 10)
print(results)
print(type(results))
```
