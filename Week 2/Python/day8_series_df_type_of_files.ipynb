```python
import pandas as pd
```

```python
result = pd.Series(["Japan", "Austria", "China", "Nepal"])
print(result)
```

```python
number_series = pd.Series([10,20,30,40],  index = ["ek", "dwe", "tin", "char"])
print(number_series)
```

```python
# NOTE:

# pd            --> pandas
# pd.Series()   --> Method
# [10,20,30,40] --> 1st Argument
# index = ["ek", "dwe", "tin", "char"] --> 2nd Argument


```

```python
# A collection of series is  known as dataframe
```

```python
# Creating a Dictionary
student_data = {
    "Name" : ["Surya","Loyas","Nilas"],
    "Age" : [20,26,30],
    "Location" : ["ktm","atm","ptn"],
    "Course" : ["Python","sql","Mern"]
}

# Creating dataframe
student_dataframe =  pd.DataFrame(student_data)
print(student_dataframe)
```
