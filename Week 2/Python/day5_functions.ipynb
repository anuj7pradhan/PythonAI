```python
# print("Hello Abis")
# print("Hello Prabi")
# print("Hello Apsar")
# print("Hello nista")
# print("Hello Abi")
# It's Repeted a lot 
# so let's introduce function

# def keyword --> 
# def greet():      --> define a function
#   prit("Hello")   --> action to be oerformed/ block of code to be executed
# 1. Function Reduces a code duplication
#       DRY (Don't Repeat Yourself)
# 2. Easier Maintinance

# now start a function

# Without parameter and arguments

def greeting():
    print("HELLO")

# Calling the function
greeting()


# Parameter and Argument
# To get different output
# Different scenario executes different output


# Parameter--> receiving placeholder
# Argument --> Input injection. 

def greet(name):        # here: name, is a Parameter
    print("Hello",name)

greet("Abis")   # here: Abis is an Argument
greet("Prabi")
greet("Apsar")
greet("nista")
greet("Abi")



```

```python
# return keyword

def add(num1,num2):
    return num1 + num2
sum = add(12,6)
print("Addition is",sum)
```

```python
# return name and age

def greet(name, age):
    return name, age
name_age = greet("Anuj",22)
print(name_age)
```

```python
# Tuple destructuring
def student():
    return "Anuj",23,"Kirtipur"
name, age, location = student()
print(name,age,location)
```
