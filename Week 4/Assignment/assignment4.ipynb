```python
# 1.⁠ ⁠Line chars
# Create line charts for: 
# Monthly student enrollments 

df['Month'] = df['EnrollmentDate'].str.slice(0, 7)

enrollment_counts = df['Month'].value_counts().sort_index()

plt.figure(figsize= (8, 8))
plt.plot(enrollment_counts.index, enrollment_counts.values, marker='o', color='red')
plt.title('Monthly Student Enrollments')
plt.xlabel('Month')
plt.ylabel('Enrollments')
plt.xticks(rotation=45)
plt.grid(True)
```

```python
# Monthly revenue trend 

monthly_revenue = df.groupby('Month')['Revenue'].sum()

plt.figure(figsize=(10, 4))
plt.plot(monthly_revenue.index, monthly_revenue.values, marker='o', color='green')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (NPR)')
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.savefig('monthly_revenue_trend.png', dpi=300)

```

```python
# Average course progress by month 
# df['Month'] = df['EnrollmentDate'].str.slice(0, 7)

avg_progress = df.groupby('Month')['CourseProgress'].mean()

plt.figure(figsize=(8,6))
plt.plot(avg_progress.index,avg_progress.values,marker ="o", color = "purple")
plt.title("Average course progress by month ")
plt.xlabel("Month")
plt.ylabel("Average progress")
plt.xticks(rotation = 45)
plt.savefig("Average course progress by month.png")
plt.grid(True)
```

```python
# Average attendance percentage by month
avg_attendance = df.groupby('Month')["AttendancePercentage"].mean()

plt.plot(avg_attendance.index, avg_attendance.values, marker = "o", color = "red")
plt.title("Average attendance percentage by month")
plt.xlabel("Months")
plt.ylabel("Average attendance percentage")
plt.xticks(rotation = 45)
plt.grid(True)
```

```python
# 2.⁠ ⁠Bar Charts
# Create bar charts showing:

# Number of students in each course
course_count = df.groupby("Course")["Course"].value_counts()
plt.bar(course_count.index, course_count.values)

plt.title("Number of students in each course")
plt.xlabel("Course")
plt.ylabel("Number of students")
plt.xticks(rotation = 45)
plt.grid(axis= "y", linestyle = "--", alpha = 0.8)
```

```python
# Revenue generated by each course

revenew_gen = df.groupby("Course")["Revenue"].sum().sort_values(ascending = False)
revenew_gen
plt.bar(revenew_gen.index,revenew_gen.values)

plt.title("Revenue generated by each course")
plt.xlabel("Course")
plt.ylabel("Revenue")
plt.xticks(rotation = 45)
plt.grid(axis= "y", linestyle = "--", alpha = 0.8)

```

```python
# Average final score by course

avg_score = df.groupby("Course")["FinalScore"].mean().sort_values(ascending = False)
plt.figure(figsize=(10,8))
plt.bar(avg_score.index,avg_score.values)


plt.title("Average final score by course")
plt.xlabel("Course")
plt.ylabel("Average Final score")
plt.xticks(rotation = 45)
plt.grid(axis= "y", linestyle = "--", alpha = 0.8)
```

```python
# Average attendance percentage by course
avg_attendance = df.groupby("Course")["AttendancePercentage"].mean().sort_values(ascending = False)
plt.bar(avg_attendance.index, avg_attendance.values)
plt.title('Average Attendance Percentage by Course', fontsize=14, fontweight='bold')
plt.xlabel('Course')
plt.ylabel('Average Attendance')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
```

```python
# Number of students from each city
city = df.groupby("City")["StudentName"].count()
plt.bar(city.index, city.values)
plt.title('Number of students from each city')
plt.xlabel('City')
plt.ylabel('Number of students')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
```

```python
# Number of students under each instructor
no_instructor = df.groupby("Instructor")["StudentName"].count()

plt.bar(no_instructor.index, no_instructor.values)
plt.title("Number of students under each instructor")
plt.xlabel("Instructor Name")
plt.ylabel("Number of students")
plt.xticks(rotation = 45)
plt.grid(alpha = 0.8, axis = "y", linestyle =  "--")
```

```python
# 3.⁠ ⁠Horizontal Bar Charts
# Create horizontal bar charts for:
# Top 10 courses by enrollment

course_counts = df['Course'].value_counts().head(10).sort_values(ascending=True)
# course_counts.index
# course_counts.values

plt.barh(course_counts.index, course_counts.values)
plt.title("Top 10 courses by enrollment")
plt.xlabel("Number of students")
plt.ylabel("Course")
plt.grid(axis='x', linestyle='--', alpha=0.7)
```

```python
# Top 10 instructors by student enrollment
std_enrollment = df['Instructor'].value_counts().head(10).sort_values(ascending=True)
# print(std_enrollment)
# std_enrollment.index
# std_enrollment.values

plt.barh(std_enrollment.index,std_enrollment.values)
plt.title("Top 10 instructors by student enrollment")
plt.xlabel("Number of students" )
plt.ylabel("Instructor")
```

```python
# Top 10 students with the highest final scores

highest_final_score = df["FinalScore"].value_counts().head(10).sort_values(ascending = True)
# highest_final_score.index
# highest_final_score.values

plt.barh(highest_final_score.index,highest_final_score.values)
plt.title("Top 10 students with the highest final scores")
plt.xlabel("Number of students" )
plt.ylabel("Highest   final score")
plt.grid(axis = "x",linestyle = '--', alpha = 0.7)
```
