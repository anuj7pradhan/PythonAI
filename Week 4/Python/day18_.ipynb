```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('daraz_dataset.csv')
```

```python
df
```

```python
# SUPPOSE    plt.hist() --> 10 bins
# 74200/10 = 74200
# 1st bin 800 = 8220
# 2nd bin 8220 = 15640
# 3rd bin 15640 = 23060
# 4th bin 23060 = 30480
```

```python
bins =[0,5000,10000,20000] # We can define our own range in bins
plt.hist(df["UnitPrice"],bins= bins, color="purple",edgecolor = "black", alpha = 0.6)
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.title("Unit Price Frequency")
```
