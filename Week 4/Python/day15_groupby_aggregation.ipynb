```python
# grouping data

df.groupby("category")
```

```python
# loop iteration of group

# when we pass the print(data), it excecutes all the groups
for category,data in df.groupby("category"):
    # print(category)
    print(data)

```

```python
# group the Electronics in separate form
# get group method to get the group of data

Electronics = df.groupby("category").get_group("Electronics")
Electronics
```

```python
# getting all the Fashion group items
 
Fashion = df.groupby("category").get_group("Fashion")
Fashion
```

```python
# single aggregation
df.groupby("category")["price"].min()
```

```python
# Multi Aggregation

df.groupby("category")["price"].agg(["max", "min", "count"])
```

```python
# cleaning the price column to numeric

df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
df.groupby("category").agg({
    "price" : "mean",
    "rating" : "min",
    "reviews" : sum
})
```
