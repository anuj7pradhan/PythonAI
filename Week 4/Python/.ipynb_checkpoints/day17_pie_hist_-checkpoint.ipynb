```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
index = [1,2,3,4,5]
price = [1000,2050,3490,3000,1550] # Vertical
plt.plot(index,price)
plt.title("INDEXING")
plt.Figure(figsize=(12,6))
# plt.figure(figsize=(6,6)) 6,6 is default size
plt.show()
```

```python
df = pd.read_csv('daraz_dataset.csv')
```

```python
df
```

```python
df.head()
```

```python
category = df.groupby("Category")["Product"].count()
plt.figure(figsize=(18,8))
plt.bar(category.index,category.values) #category.index = X-axis ,category.values = Y-axis)
plt.title("Products and Category Bar Graph")
plt.xlabel("Categories",fontsize = 20)
plt.ylabel("Products",fontsize = 20)
plt.grid(axis="y",linestyle="--",alpha=1)   # alpha is 0-1 range

```

```python
category
```

```python
colors = [
    "green",
    "blue",
    "red",
    "pink",
    "orange",
    "grey",
    "purple"
]
plt.figure(figsize=(14,7))
plt.pie(category.values, labels=category.index, colors = colors, shadow= True, startangle= 90, explode=(0.1,0.2,0,0,0,0,0))
```

```python
gender = df.groupby("Gender")["Product"].count()
```

```python
gender
```

```python
plt.pie(gender.values, labels= gender.index)
plt.title("Gender wise Products")
```

```python
# Histogram
```

```python
plt.hist(df["UnitPrice"], color="purple",edgecolor = "black", alpha = 0.6)
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.title("Unit Price Frequency")
```

```python
plt.hist(df["Category"], color="blue",edgecolor = "black", alpha = 0.1)
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.title("Distribution of Category")
```

```python
plt.hist(df["PaymentMethod"], color="red",edgecolor = "black", alpha = 0.3)
plt.xlabel("Payement Method")
plt.ylabel("Frequency")
plt.title("Distributuion of Payment Method")
```
