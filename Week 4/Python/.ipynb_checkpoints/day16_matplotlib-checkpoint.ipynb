```python
df = pd.read_csv('daraz_dataset.csv')
```

```python
df
```

```python
df.head()
```

```python
category = df.groupby("Category")["Product"].count()
plt.figure(figsize=(18,8))
plt.bar(category.index,category.values) #category.index = X-axis ,category.values = Y-axis)
plt.title("Products and Category Bar Graph")
plt.xlabel("Categories",fontsize = 20)
plt.ylabel("Products",fontsize = 20)
plt.grid(axis="y",linestyle="--",alpha=1)   # alpha is 0-1 range

```
