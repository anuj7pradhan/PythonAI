```python
numeric_columns = [
    'Age',
    'Quantity',
    'UnitPrice', 
    'DiscountPercent',
    'TotalAmount',
    'Rating'
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column],errors="coerce")
```

```python
df["Quantity"]  = df["Quantity"].fillna(df["Quantity"].mean())
```
