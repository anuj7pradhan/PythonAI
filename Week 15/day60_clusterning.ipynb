```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("digital_pathshala_student_clustering.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
numerical_columns = [
    "StudyHoursPerWeek",
    "AttendancePercentage",
    "AssignmentCompletionPercentage"
]
```

```python
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
X = df[[
     "StudyHoursPerWeek",
    "AttendancePercentage",
    "AssignmentCompletionPercentage"
]]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
k_means = KMeans(
    n_clusters = 3, # no. of cluster
    init = "k-means++", # starting centroid 
    n_init = 10,
    random_state= 42,
    tol= 0.0001
)
k_means
```

```python
df["Group"] = k_means.fit_predict(X_scaled)
df
```

```python
df["Group"] = df["Group"].replace({
    0: "A",
    1: "B",
    2: "C"
})
```
