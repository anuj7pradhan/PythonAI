```python
# 3. Outlier Detection using iqr
# 
```

```python
# For the following columns:
# Course Fee

# course_fee_q1 (cq1)
cq1 = df["course_fee"].quantile(0.25)
cq1
```

```python
# course_fee_q2 (cq2)

cq2 = df["course_fee"].quantile(0.5)
cq2
```

```python
# course_fee_q3 (cq3)

cq3 = df["course_fee"].quantile(0.75)
cq3
```

```python
# course_fee_iqr (ciqr)

ciqr = cq3 - cq1
ciqr
```

```python
# course_fee_lower-bound (clb)

clb = cq1 - 1.5 * ciqr
clb
```

```python
# course_fee_upper-bound (cub)

cub = cq3 + 1.5 * ciqr
cub
```

```python
# course_fee_outliers (co)

co = df[(df["course_fee"] < clb) | (df["course_fee"] > cub)]
co
```

```python
course_fee_median = df["course_fee"].median()
course_fee_median
```

```python
# now let's replace outlier by median

df.loc[(df["course_fee"] < clb) | (df["course_fee"] > cub),"course_fee"] = course_fee_median
df
```

```python
# Attendance Percentage

# attendance_percent_q1 (aq1)

aq1 = df["attendance_percent"].quantile(0.25)
aq1
```

```python
# attendance_percent_q2 (aq2)

aq2 = df["attendance_percent"].quantile(0.5)
aq2
```

```python
# attendance_percent_q3 (aq3)

aq3 = df["attendance_percent"].quantile(0.75)
aq3
```

```python
# attendance_percentage_iqr (aiqr)

aiqr = aq3 - aq1
aiqr
```

```python
# attendance_percentage_lower_bound (a_lower_bound)

a_lower_bound = aq1 - 1.5 * aiqr
a_lower_bound
```

```python
# attendance_percentage_upper_bound (a_upper_bound)

a_upper_bound = aq3 + 1.5 * aiqr
a_upper_bound
```

```python
# attendance_percentage_outlier (ao)

ao = df[(df["attendance_percent"] < a_lower_bound) | (df["attendance_percent"] > a_upper_bound)]
ao
```

```python
# As we didn't get any outliers on attendance_percentage, just skip to next
```
