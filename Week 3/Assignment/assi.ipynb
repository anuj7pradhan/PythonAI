```python
# Clean the dataset by performing the following tasks:
```

```python
# Identify and remove duplicate records.
df.drop_duplicates(inplace = True)
```

```python
df
```

```python
# Detect and handle missing values using appropriate techniques.
df.isnull().sum()
```

```python
# convert incorrect data types into their appropriate formats.
print(df.dtypes)
df["age"] = pd.to_numeric(df["age"], errors = "coerce")
df["course_fee"] = pd.to_numeric(df["course_fee"], errors = "coerce")
df["discount"] = pd.to_numeric(df.discount, errors='coerce')
df["assignment_score"] = pd.to_numeric(df["assignment_score"], errors='coerce')
df["final_score"] = pd.to_numeric(df["final_score"], errors='coerce')
df["rating"] = pd.to_numeric(df["rating"], errors='coerce')


```

```python
# Standardize inconsistent values such as city names, course names, and other text-based columns

# df["student_name, course, city, gender, payment_status, mentor, email, completion_status"] = df["student_name, course, city, gender, payment_status, mentor, email, completion_status"].str.lower()

df["student_name"] = df["student_name"].str.title()
```
