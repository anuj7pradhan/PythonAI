```python
# QUARTILES ---> Quarter division

    # Q1 --> 25% below
    # Q2 --> Median/ below 50% , Above 50%
    # Q3 --> Below 75% , Above 75%
```

```python
salary = np.array([1000,2000,3000,5000,7000,8000,9000,12000])
# Q2
np.median(salary)
```

```python
# Q1
np.percentile(salary,25)
```

```python
#  Q1
np.quantile(salary,0.25)
```

```python
# Q3
np.percentile(salary,75)
```

```python
#  Q3
np.quantile(salary,0.75)
```

```python
# IQR method
# IQR = Q3 - Q1 =  8250.0 -  2750.0  = 5500.0
```

```python
# 1. lower bound = Q1 - 1.5  * IQR =  -5500.0

# 2. Upper bound = Q3 + 1.5 * IQR = 16500.0
```

```python
import pandas as pd
spd = pd.DataFrame({
    "salary" : [1000,2000,50000,3000,5000,7000,8000,9000,12000,100000]
})
```

```python
spd
```

```python
q2 = spd["salary"].quantile(0.5)
q2
```

```python
q1 = spd["salary"].quantile(0.25)
q1
```

```python
q3 = spd["salary"].quantile(0.75)
q3
```

```python
iqr = q3 - q1
iqr
```

```python
# lower_bound
lb = q1 - 1.5 * iqr
lb
```

```python
# Upper_bound
ub = q3 + 1.5 * iqr
ub
```

```python
outlier = spd[(spd["salary"] < lb) | (spd["salary"] > ub)]
outlier
```
