```python
# QUARTILES ---> Quarter division

    # Q1 --> 25% below
    # Q2 --> Median/ below 50% , Above 50%
    # Q3 --> Below 75% , Above 75%
```

```python
salary = np.array([1000,2000,3000,5000,7000,8000,9000,12000])
# Q2
np.median(salary)
```

```python
# Q1
np.percentile(salary,25)
```

```python
#  Q1
np.quantile(salary,0.25)
```

```python
# Q3
np.percentile(salary,75)
```

```python
#  Q3
np.quantile(salary,0.75)
```

```python
# IQR method
# IQR = Q3 - Q1 =  8250.0 -  2750.0  = 5500.0
```

```python
# 1. lower bound = Q1 - 1.5  * IQR =  -5500.0

# 2. Upper bound = Q3 + 1.5 * IQR = 16500.0
```

```python
8250 + 1.5 *  5500
```
