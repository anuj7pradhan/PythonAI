```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Salary": [1000,2000,3000,4000,5000,6000,100000]
})
```

```python
df
```

```python
q1 = df["Salary"].quantile(0.25)
q1
```

```python
q2 = df["Salary"].mean()
q2
```

```python
q3 = df["Salary"].quantile(0.75)
q3
```

```python
iqr = q3 - q1
iqr
```

```python
lower_bound = q1 - 1.5 * iqr
lower_bound
```

```python
upper_bound = q3 + 1.5 * iqr
upper_bound
```

```python
outliers = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]
outliers
```

```python
ddf = pd.read_csv("bad_daraz_dataset.csv")
```

```python
ddf
```

```python
# Replacing price data to numeric
ddf["price"] = pd.to_numeric(ddf["price"],errors = "coerce")
```

```python
ddf
```

```python
# Outlier detcting

# Quartile Q3, Q1 and iqr

```

```python
q1 = ddf["price"].quantile(0.25)
q1
```

```python
q2 = ddf["price"].quantile(0.50)
q2
```

```python
q3 = ddf["price"].quantile(0.75)
q3
```

```python
iqr = q3 - q1
iqr
```

```python
lower_bound = q1 - 1.5 * iqr
lower_bound
```

```python
upper_bound = q3 + 1.5 * iqr
upper_bound
```

```python
outliers = ddf[(ddf["price"] < lower_bound) | (ddf["price"] > upper_bound)]
outliers
```

```python
median = ddf["price"].median()
median
```

```python
# detected outlier change into lower_bound or upper_bound
    # Since the detected value (75000.0) is greater than the upper_bound (10575.0) 
    # We gotta change it to upper_bound value i.e 10575.0
# here the clip() method to fix the outliers

ddf["price"]=ddf["price"].clip(lower_bound,upper_bound)
```
