```python
# Classification types
    # 1. Will customer buy life insurance??             i.e   Binary Classification
    # 2. Which party a person is going to vote for??    i.e   Multiclass Classification
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
```

```python
df = pd.DataFrame({
    "age" :[22,25,47,52,46,56,55,60,62,61,18,28,27,29,49],
    "have_insurance":[0,0,1,0,1,1,0,1,1,1,0,0,0,0,1]
})
```

```python
df
```

```python
plt.scatter(df["age"], df["have_insurance"],color='blue', alpha=0.7, edgecolors='k', s=80)
plt.yticks([0, 1], ['No (0)', 'Yes (1)'])
plt.xlabel('Age')
plt.ylabel('Have Insurance')
plt.title('Relationship Between Age and Insurance Status')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
```

```python
# Prepare features and target
X = df[["age"]]
y = df["have_insurance"]

```

```python
# Fit logistic regression model
model = LogisticRegression()
model.fit(X, y)

# Generate values for the smooth sigmoid curve
X_test = np.linspace(df["age"].min(), df["age"].max(), 300).reshape(-1, 1)
y_prob = model.predict_proba(X_test)[:, 1]  # Probability of class 1

```

```python
# Plotting
plt.figure(figsize=(8, 5))
plt.scatter(
    df["age"],
    df["have_insurance"],
    color="royalblue",
    edgecolors="black",
    s=100,
    alpha=0.8,
    label="Actual Data",
)
plt.plot(
    X_test,
    y_prob,
    color="crimson",
    linewidth=2.5,
    label="Sigmoid Curve (Logistic Regression)",
)


# Customize plot
plt.yticks([0, 0.5, 1], ["0 (No)", "0.5", "1 (Yes)"])
plt.xlabel("Age", fontsize=12)
plt.ylabel("Probability of Having Insurance", fontsize=12)
plt.title(
    "Sigmoid Function Fitted to Insurance Data", fontsize=14, fontweight="bold"
)


plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()

# Save and display
plt.savefig("sigmoid_curve.png", dpi=300)
```
