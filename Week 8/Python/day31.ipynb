```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
# inconsistent data cha bhane polynomial lagu huncha
df =  pd.read_csv("student_multiple_features.csv")
```

```python
df
```

```python
df["Attendance"] = pd.to_numeric(df["Attendance"], errors= "coerce")
df["PreviousMarks"] = pd.to_numeric(df["PreviousMarks"], errors= "coerce")

df = df.dropna()
```

```python
df
```

```python
# Features and Labels separation

X = df[[
    "StudyHours",
    "Attendance",
    "PreviousMarks"	
]]

Y = df["FinalMarks"]
```

```python
X = X.to_numpy()
Y = Y.to_numpy()
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python

poly = PolynomialFeatures(
    degree = 2,
    include_bias= False
)
poly
```

```python
poly_x_train = poly.fit_transform(X_train)
poly_x_train
```

```python
poly_x_test = poly.fit_transform(X_test)
poly_x_test
```

```python
model = LinearRegression()
model.fit(poly_x_train,Y_train)
```

```python
prediction = model.predict(poly_x_test)
prediction
```

```python
comparision = pd.DataFrame({
    "actual": Y_test,
    "Prediction" : prediction
})
```

```python
comparision
```

```python
mae = mean_absolute_error(
    Y_test,
    prediction
)
mae
```

```python
mse = mean_squared_error(
    Y_test,
    prediction
)
mse
```

```python
rmse = np.sqrt(mse)
print(rmse)
```
